## 步驟 2：啟動 3D 果蠅互動模擬器 (可直接在下方操作、觀看飛行與錄製 3D 畫面)

```python
from IPython.display import display, HTML
import html

# 讀取 Web 視圖與已打包的 3D 動力學 Bundle (內含 Three.js + 完整神經網絡與關節運動學)
with open('colab/index.html', 'r', encoding='utf-8') as f:
    html_template = f.read()

with open('colab/desktop_fly_bundle.js', 'r', encoding='utf-8') as f:
    js_bundle = f.read()

# 將 JS bundle 嵌入至 HTML，免除 Colab 外部伺服器依賴，保證離線與極速載入
embedded_html = html_template.replace(
    '<script type="module" src="./desktop_fly_bundle.js"></script>',
    f'<script type="module">\n{js_bundle}\n</script>'
)

# 以安全 iframe 嵌入至 Colab 儲存格中
escaped_doc = html.escape(embedded_html, quote=True)
display(HTML(f'''
<iframe srcdoc="{escaped_doc}" width="100%" height="960px" style="border:1px solid #233354; border-radius:12px; box-shadow:0 8px 30px rgba(0,0,0,0.6);"></iframe>
'''))
```

## 步驟 3：播放與檢查錄製完成的 3D 影片

> 💡 **使用說明**：
> 1. 在上方 3D 模擬器中點擊 **「🚀 起飛飛行」** 或 **「✈️ 連續巡航飛行」**，果蠅會立即展翅在 3D 空間中翱翔！
